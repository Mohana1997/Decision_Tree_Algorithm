```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
```

```python
claimantsdata=pd.read_csv("claimants.csv")
claimantsdata
```

```python
claimantsdata.isnull().sum()
```

```python
claimantsdata.dropna(inplace = True)
```

```python
claimantsdata.isnull().sum()
```

```python
x=claimantsdata.drop(labels=["CASENUM","ATTORNEY"],axis=1)
```

```python
y=claimantsdata["ATTORNEY"]
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.20,shuffle=True,random_state=67)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
## dt_model - OVERFITTING (max depth not used)(PERFORMED GridSearchCV at last)
## dt1_model - GENERALIZED (max depth used)
```

#### MODEL 1 - dt_model [OVERFITTING (max depth not used)(PERFORMED GridSearchCV at last)]

```python
dt_model=DecisionTreeClassifier()
```

```python
dt_model.fit(x_train,y_train)
```

```python
y_train_pred = dt_model.predict(x_train)
y_train_pred
```

```python
y_train
```

```python
y_test_pred=dt_model.predict(x_test)
y_test_pred
```

```python
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix,roc_auc_score,roc_curve
```

```python
print("train_accuracy:",accuracy_score(y_train,y_train_pred))
print("test_accuracy:",accuracy_score(y_test,y_test_pred))
```

```python
print(confusion_matrix(y_train,y_train_pred))
print(confusion_matrix(y_test,y_test_pred))
```

```python
print(classification_report(y_train,y_train_pred))
```

```python
print(classification_report(y_test,y_test_pred))
```

```python
fpr, tpr, thresholds = roc_curve(y_train,dt_model.predict_proba (x_train)[:,1])

auc = roc_auc_score(y_train,y_train_pred)
print(auc)

import matplotlib.pyplot as plt
plt.plot(fpr, tpr, color='red', label='logit model ( area  = %0.2f)'%auc)
plt.plot([0, 1], [0, 1], 'k--')
plt.xlabel('False Positive Rate or [1 - True Negative Rate]')
plt.ylabel('True Positive Rate')
```

```python
fpr, tpr, thresholds = roc_curve(y_test,dt_model.predict_proba (x_test)[:,1])

auc = roc_auc_score(y_test,y_test_pred)
print(auc)

import matplotlib.pyplot as plt
plt.plot(fpr, tpr, color='red', label='logit model ( area  = %0.2f)'%auc)
plt.plot([0, 1], [0, 1], 'k--')
plt.xlabel('False Positive Rate or [1 - True Negative Rate]')
plt.ylabel('True Positive Rate')
```

```python
#has low bias(low error in training data) and high variance(high error in test data) - this model is OVERFITTING
```

```python
## so we introduce max_depth value so that the tree is not grown more than the level we have mentioned 
## we have purposefully decided to cut the tree
```

```python
## WE always want an generalized model
```

# 3 Model Validation Techniques

### 1) Train test split

### 2) KFold CV

```python
from sklearn.model_selection import KFold,cross_val_score

kfold = KFold(n_splits=5,shuffle=True, random_state=123)

cv_scores = cross_val_score(estimator = dt_model,X = x,y=y,cv=kfold)

print(cv_scores)
print("Overall Accuracy:", cv_scores.mean().round(2))
print("Overall Std.Deviation:", cv_scores.std().round(2))
```

### 3) LOOCV

```python
from sklearn.model_selection import LeaveOneOut
loocv = LeaveOneOut()

cv_scores = cross_val_score(estimator = dt_model,X = x,y=y,cv=loocv)

print(cv_scores)
print("Overall Accuracy:", cv_scores.mean().round(2))
print("Overall Std.Deviation:", cv_scores.std().round(2))
```
